```python
from sqlalchemy import create_engine
import pandas as pd

# Example for MySQL
engine = create_engine("mysql+pymysql://root:Amulya%40123@localhost:3306/retail_orders")
conn=engine.connect()
# Write DataFrame into SQL
df.to_sql("df_orders", con=engine, if_exists="replace", index=False)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

#  MySQL query
query = """
SELECT product_id, SUM(sale_price) AS sales
FROM df_orders
GROUP BY product_id
ORDER BY sales DESC
LIMIT 10;
"""

# Run query and fetch results into Pandas
df_top10 = pd.read_sql(query, conn)
print(df_top10)

# Visualization
plt.figure(figsize=(10,6))
plt.bar(df_top10['product_id'], df_top10['sales'])
plt.xlabel("Product ID")
plt.ylabel("Sales Revenue")
plt.title("Top 10 Highest Revenue Products")
plt.xticks(rotation=45)
plt.show()
```

```python
#mysql Query
#2.find top 5 highest selling products in each region
query='''select region,product_id,sales
 from(
 select region,product_id,sum(sale_price) as sales,
          row_number() over (partition by region order by sum(sale_price) desc) as rn
from df_orders
group by region,product_id
)ranked
where rn<=5;'''
df_top5_region=pd.read_sql(query,conn)
print(df_top5_region)
# Visualization
plt.figure(figsize=(12,8))

# Loop through each region to plot separately
for region in df_top5_region['region'].unique():
    subset = df_top5_region[df_top5_region['region'] == region]
    plt.bar(subset['product_id'], subset['sales'], label=region)

plt.xlabel("Product ID")
plt.ylabel("Sales Revenue")
plt.title("Top 5 Highest Selling Products in Each Region")
plt.xticks(rotation=45)
plt.legend(title="Region")
plt.show()
```

```python

#sql query
#3.find month over month growth comparison for 2022 and 2023 sales eg: jan 2022 vs jan 2023

query='''
with cte as (
select year(order_date) as order_year,month(order_date) as order_month,
sum(sale_price) as sales
from df_orders
group by year(order_date),month(order_date)
-- order by year(order_date),month(order_date)
	)
select order_month
, sum(case when order_year=2022 then sales else 0 end) as sales_2022
, sum(case when order_year=2023 then sales else 0 end) as sales_2023
from cte 
group by order_month
order by order_month;'''
#visualization
df_mom=pd.read_sql(query,conn)
print(df_mom)
plt.figure(figsize=(10,6))
plt.plot(df_mom['order_month'], df_mom['sales_2022'], marker='o', label='2022')
plt.plot(df_mom['order_month'], df_mom['sales_2023'], marker='o', label='2023')

plt.xticks(df_mom['order_month'], 
           ['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec'])
plt.xlabel("Month")
plt.ylabel("Sales Revenue")
plt.title("Month-over-Month Sales Comparison (2022 vs 2023)")
plt.legend()
plt.grid(True)
plt.show()
```

```python
#4.for each category which month had highest sales
query='''
with cte as(
 select category,date_format(order_date,'%%Y-%%m') as order_year
 ,sum(sale_price) as sales from df_orders
 group by category,date_format(order_date,'%%Y-%%m')

 )
 Select *from (
 select *,
 row_number() over (partition by category order by sales desc) as rn
 from cte
 )a
where rn=1;
'''

#connection
df_result=pd.read_sql(query,conn)
print(df_result)
#Visualization
plt.figure(figsize=(8,6))
plt.bar(df_result['category'], df_result['sales'], color='skyblue')

# Annotate sales values + month
for i, (val, month) in enumerate(zip(df_result['sales'], df_result['order_year'])):
    plt.text(i, val, f"{val}\n({month})", ha='center', va='bottom')

plt.title("Highest Sales Month per Category")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.show()


```

```python
 #5.which sub category had highest growth by profit in 2023 compare to 2022
query=''' 
with cte as(
 select sub_category,year(order_date) as order_year,sum(sale_price) as sales
 from df_orders
 group by sub_category,year(order_date)
 order by sub_category,year(order_date)
 )
 , cte2 as(
 select  sub_category,
 sum(case when order_year=2022 then sales  else 0 end) as sales_2022,
 sum(case when order_year=2023 then sales else 0 end) as sales_2023
 from cte
 group by sub_category
 )
 select *,
 (sales_2023-sales_2022)*100/sales_2022  as total
 from cte2 
 order by total desc
 limit 1;'''
df_growth=pd.read_sql(query,conn)
print(df_growth)
#visualization
plt.figure(figsize=(8,6))
plt.bar(df_growth['sub_category'], df_growth['total'], color='green')
plt.ylabel("Growth %")
plt.title("Highest Growth Sub-Category (2023 vs 2022)")
plt.xticks(rotation=45)
plt.show()
```
